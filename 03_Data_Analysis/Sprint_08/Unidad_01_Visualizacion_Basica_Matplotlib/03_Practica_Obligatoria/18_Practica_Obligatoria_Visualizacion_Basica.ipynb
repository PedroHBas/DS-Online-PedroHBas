#### #1.2  

Recrea la siguiente figura, para ello construye en apartados separados cada gráfico y luego haz un apartado en el que se junten los cuatro.  

Las figuras son:
1. Las ciudades más pobladas, con los numeros de las poblaciones (tendrás que usar texto y recordar que para centrarlos se necesita poner el argumento `ha` a "center"). (Colores: azul, gris, aceituna, marrón, con trasparencia)  

2. Histograma del área total ocupada por las ciudades. (50 bins, fuente = "Bradley Hand ITC", tamaño 18, color azul acero o "steelblue")
3. El scatter de área de agua de un ciudad con su poblacion.
4. El boxplot de la altura de las ciudades.

```python
df1 = df.sort_values('population_total', ascending=False).iloc[:5]

ax1 = plt.axes()
ax1.set_title('Ciudades más pobladas de California')
ax1.bar(df1.city.transform(lambda x: ''.join([' ' + letra if letra.isupper() and letra != x[0] else letra for letra in x])), df1.population_total / 1000, color=['blue', 'gray', 'red', 'olive', 'brown'], alpha=0.5)
ax1.set_xlabel('Ciudades')
ax1.set_ylim(0, 5000)
ax1.set_ylabel('Habitantes(miles)')
for i, v in enumerate(df1.population_total / 1000):
    ax1.text(i, v + 100, str(v), ha='center')
```

```python
ax2 = plt.axes()
ax2.hist(df.area_total_km2, bins=50, color='steelblue', label='Hist Area', alpha=0.7)
ax2.set_xlabel('Superficie ($km^2$)', font='Comic Sans MS')
ax2.set_ylabel('Número de Ciudades', font='Comic Sans MS')
ax2.legend();
```

```python
df.sort_values('area_water_km2', ascending=False).iloc[:2]
```

```python
ax3 = plt.axes()
ax3.scatter(df.area_water_km2, df.population_total / 1000, c='g', alpha=0.5)
ax3.set_xlabel('Área de agua en la ciudad ($km^2$)')
ax3.set_ylabel('Población (miles habitantes)')
ax3.annotate('Outliners', (88, 3884),(60, 3500), arrowprops=dict(arrowstyle='->'))
ax3.annotate('         ', (122.27, 1345),(60, 3500), arrowprops=dict(arrowstyle='->'))
ax3.set_title('Población por kilómetro cuadrado de agua')
```

```python
ax4 = plt.axes()
ax4.boxplot(df.elevation_m, orientation='horizontal', patch_artist=True,
            boxprops=dict(facecolor='red', ),
            medianprops=dict(color='black'))
ax4.set_xlabel('Altitud (m)')
ax4.set_title('Diagrama de caja de la Altitud de las ciudades de California')
```

```python
fig, ax = plt.subplots(2, 2, figsize=(20, 20))

ax[0,0].set_title('Ciudades más pobladas de California')
ax[0,0].bar(df1.city.transform(lambda x: ''.join([' ' + letra if letra.isupper() and letra != x[0] else letra for letra in x])), df1.population_total / 1000, color=['blue', 'gray', 'red', 'olive', 'brown'], alpha=0.5)
ax[0,0].set_xlabel('Ciudades')
ax[0,0].set_ylim(0, 5000)
ax[0,0].set_ylabel('Habitantes(miles)')
for i, v in enumerate(df1.population_total / 1000):
    ax[0,0].text(i, v + 100, str(v), ha='center')

ax[0,1].hist(df.area_total_km2, bins=50, color='steelblue', label='Hist Area', alpha=0.7)
ax[0,1].set_xlabel('Superficie ($km^2$)', font='Comic Sans MS')
ax[0,1].set_ylabel('Número de Ciudades', font='Comic Sans MS')
ax[0,1].legend()

ax[1,0].scatter(df.area_water_km2, df.population_total / 1000, c='g', alpha=0.5)
ax[1,0].set_xlabel('Área de agua en la ciudad ($km^2$)')
ax[1,0].set_ylabel('Población (miles habitantes)')
ax[1,0].annotate('Outliners', (88, 3884),(60, 3500), arrowprops=dict(arrowstyle='->'))
ax[1,0].annotate('         ', (122.27, 1345),(60, 3500), arrowprops=dict(arrowstyle='->'))
ax[1,0].set_title('Población por kilómetro cuadrado de agua')

ax[1,1].boxplot(df.elevation_m, orientation='horizontal', patch_artist=True,
            boxprops=dict(facecolor='red', ),
            medianprops=dict(color='black'))
ax[1,1].set_xlabel('Altitud (m)')
ax[1,1].set_title('Diagrama de caja de la Altitud de las ciudades de California');
```
